## Construcción aglomerativa y enlaces
Comenzamos con un grupo por observación y fusionamos dos grupos en cada paso.
Simple usa la distancia mínima entre pares; completo, la máxima; promedio, la media de todas las distancias
entre pares. Ward minimiza el incremento de suma de cuadrados:
$$\Delta(A,B)=\frac{n_A n_B}{n_A+n_B}\|\mu_A-\mu_B\|^2.$$
La identidad se obtiene al descomponer la suma de cuadrados respecto de la media conjunta.
En SciPy, la altura de enlace Ward es $\sqrt{2\Delta}$, no $\Delta$.
El enfoque divisivo empieza con todos los puntos y divide; lo estudiamos conceptualmente.

### Para resolver

Para A={0,2} y B={5}, calcule el incremento de Ward y la altura que mostraría SciPy para esa fusión.

**Mi respuesta y evidencia:**

Calculamos el incremento de la suma de cuadrados del criterio de Ward ($\Delta(A, B)$) y la correspondiente altura de enlace que reporta SciPy para la fusión de los grupos $A = \{0, 2\}$ y $B = \{5\}$:

### 1. Estadísticos de cada grupo
- Grupo $A$: $n_A = 2$, $\mu_A = \frac{0 + 2}{2} = 1.0$.
  $SSE(A) = (0 - 1)^2 + (2 - 1)^2 = 1 + 1 = 2.0$.
- Grupo $B$: $n_B = 1$, $\mu_B = 5.0$.
  $SSE(B) = (5 - 5)^2 = 0.0$.
  $SSE_{\text{antes}} = SSE(A) + SSE(B) = 2.0 + 0.0 = 2.0$.

### 2. Media y SSE del grupo fusionado $A \cup B$
- Puntos fusionados: $\{0, 2, 5\}$, tamaño $n_{AB} = 3$.
  Media conjunta: $\mu_{AB} = \frac{0 + 2 + 5}{3} = \frac{7}{3} \approx 2.333333$.
- Suma de cuadrados después de la fusión:
  $SSE(A \cup B) = \left(0 - \frac{7}{3}\right)^2 + \left(2 - \frac{7}{3}\right)^2 + \left(5 - \frac{7}{3}\right)^2 = \frac{49}{9} + \frac{1}{9} + \frac{64}{9} = \frac{114}{9} = \frac{38}{3} \approx 12.666667$.

### 3. Incremento de Ward ($\Delta(A, B)$)
El incremento en la inercia total es:
$$\Delta(A, B) = SSE(A \cup B) - SSE(A) - SSE(B) = \frac{38}{3} - 2 = \frac{32}{3} \approx \mathbf{10.666667}\text{ km}^2$$
*Verificación por la fórmula directa ponderada de Ward:*
$$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2 = \frac{2 \times 1}{2 + 1} |1.0 - 5.0|^2 = \frac{2}{3} \times (-4)^2 = \frac{2}{3} \times 16 = \frac{\mathbf{32}}{\mathbf{3}}\text{ km}^2$$

### 4. Altura en la matriz de enlace de SciPy (`linkage`)
La implementación de `scipy.cluster.hierarchy.linkage` con `method='ward'` define la distancia de fusión en términos de la métrica euclidiana generalizada:
$$d_{\text{SciPy}}(A, B) = \sqrt{2 \Delta(A, B)}$$
Sustituyendo el valor analítico:
$$d_{\text{SciPy}}(A, B) = \sqrt{2 \times \frac{32}{3}} = \sqrt{\frac{64}{3}} = \frac{8}{\sqrt{3}} \approx \mathbf{4.618802}\text{ km}$$
*(Verificado con el assert en Celda 7: `assert np.isclose(ward_toy[-1,2], np.sqrt(64/3))`).*

- **Unidades:** Incremento de inercia en $\text{km}^2$; altura de enlace de SciPy en $\text{km}$.
- **Limitación:** El criterio de Ward está estrictamente acoplado a la descomposición de varianza euclidiana en $\mathbb{R}^d$.

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
small=load_data('jerarquico'); Z=small[['x_km','y_km']].to_numpy()
fig,axes=plt.subplots(2,2,figsize=(11,7))
for ax,method in zip(axes.ravel(),['single','complete','average','ward']):
    tree=linkage(Z,method=method,metric='euclidean')
    dendrogram(tree,truncate_mode='lastp',p=18,ax=ax,no_labels=True)
    ax.set(title=method,xlabel='Ramas truncadas',ylabel='Altura de enlace')
fig.tight_layout(); savefig(fig,'16_dendrogramas')
```

## Cortar el árbol y leer la partición
La distancia vertical mide el criterio de enlace. La posición horizontal es un orden de hojas, no una distancia
espacial. Un corte por altura no equivale al mismo número de grupos en todos los enlaces.
`maxclust` solicita a lo sumo K grupos; compruebe cuántos produce realmente si hay empates.

```python
tree=linkage(Z,method='ward',metric='euclidean')
rows=[]
fig,axes=plt.subplots(1,3,figsize=(12,4.5))
for ax,k in zip(axes,[2,4,8]):
    labels=fcluster(tree,t=k,criterion='maxclust')
    rows.append(dict(requested_k=k,actual_k=len(np.unique(labels)),silhouette=safe_silhouette(Z,labels)))
    ax.scatter(*Z.T,c=labels,cmap='tab10',s=12)
    spatial_axes(ax,X,f'Ward · hasta {k} grupos')
fig.tight_layout(); savefig(fig,'17_ward_espacial')
display(pd.DataFrame(rows)); save_table(pd.DataFrame(rows),'jerarquico_cortes')
ward_toy=linkage(np.array([0.,2.,5.])[:,None],method='ward')
assert np.isclose(ward_toy[-1,2],np.sqrt(64/3))
print('Fusiones del ejemplo manual:',ward_toy)
```

### Para resolver

¿Por qué enlace simple produce encadenamiento? ¿Por qué no usar Ward con haversine? ¿Es comparable la silhouette de estos 500 puntos con la de 6000?

**Mi respuesta y evidencia:**

1. **¿Por qué el enlace simple produce encadenamiento (chaining effect)?**
   El enlace simple (*single linkage*) define la distancia entre conglomerados como la distancia mínima entre cualquier par de puntos: $D(A, B) = \min_{u \in A, v \in B} d(u, v)$. Si existen observaciones aisladas o ruidosas que formen un puente tenue entre dos concentraciones densas, el algoritmo fusionará ambos grupos secuencialmente a través de ese puente, generando conglomerados alargados y serpentinos en vez de agrupamientos compactos.

2. **¿Por qué no usar Ward con distancia Haversine?**
   El método de Ward se deriva formalmente a partir del teorema de descomposición de suma de cuadrados euclidiana (teorema de Huygens): $\mathrm{SSE}(A \cup B) = \mathrm{SSE}(A) + \mathrm{SSE}(B) + \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$. Esta descomposición exige un espacio afín euclidiano $\mathbb{R}^d$. Haversine es una métrica sobre la esfera terrestre $S^2$; sobre una superficie no euclidiana, la media aritmética de coordenadas no es el minimizador de distancias al cuadrado y la identidad de Ward pierde toda validez matemática.

3. **¿Es comparable el Silhouette de estos 500 puntos con el de 6,000 puntos?**
   **NO son directamente comparables.** El coeficiente de Silhouette $s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}$ depende fuertemente de la densidad y la escala muestral local. En una submuestra reducida ($n=500$), los puntos están mucho más separados en promedio, modificando artificialmente las distancias medias intra e inter-clúster. Para una comparación experimental controlada, los métodos deben evaluarse exactamente sobre el mismo conjunto de datos, mismo soporte y misma métrica.

- **Unidades:** Distancias en $\text{km}$, Silhouette adimensional en $[-1, 1]$.
- **Limitación:** El clustering jerárquico aglomerativo tiene costo computacional $\mathcal{O}(n^2)$ en memoria y $\mathcal{O}(n^3)$ en tiempo, restringiendo su aplicación a submuestras reducidas como la de 500 observaciones.

```python
from sklearn.cluster import KMeans
km=KMeans(4,n_init=10,random_state=42).fit_predict(Z)
ward=fcluster(tree,t=4,criterion='maxclust')
print('Misma submuestra, silhouette K-means:',safe_silhouette(Z,km))
print('Misma submuestra, silhouette Ward:',safe_silhouette(Z,ward))
```
